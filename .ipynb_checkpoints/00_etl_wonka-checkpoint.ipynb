## Drop unnecessary columns

```python
df = df.loc[:, ~df.columns.str.contains("^Unnamed")]
```

```python
df = df.drop(columns=[
    "Latitude",
    "Longitude",
    "Postal Code",
    "Ship Mode"
])
```

```python
df.head()
```

```python
df=df.drop_duplicates()
```

```python
df.info()
```

```python
df.isnull().sum()
```

## Normalize column names

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("/", "_")
)
```

```python
df.head()
```

## Make Order date and Ship date a date type

```python
df["order_date"] = pd.to_datetime(df["order_date"])
df["ship_date"] = pd.to_datetime(df["ship_date"])
```

```python
df.info()
```

## Numeric values

```python
numeric_cols = ["sales", "units", "gross_profit", "cost"]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
df.info()
```

## Order date vs Ship date GAP detected

### Diference in years of order date vs ship date

```python
df["shipping_gap_days"] = (df["ship_date"] - df["order_date"]).dt.days
```

```python
df["shipping_gap_days"].describe()
```

```python
gap_median = df["shipping_gap_days"].median()
gap_median
```

Mean of difference was 2003 days = 5.5 years

### Rest the median difference of days to ship day

```python
df["ship_date"] = df["ship_date"] - pd.to_timedelta(gap_median, unit="D")
```

```python
df.loc[df["ship_date"] < df["order_date"], "ship_date"] = df["order_date"]
```

```python
(df["ship_date"] - df["order_date"]).dt.days.describe()
```

A systematic anomaly was detected in the dataset where shipping dates were consistently ~2000 days after order dates.
By analyzing the median gap (2004 days), we corrected this offset and validated the fix by ensuring the final shipping delays were within a realistic range (0–7 days).

# Upload COLI Scores CSV Dataset

```python
coli_df = pd.read_csv("coli-states-2023-2024-2025.csv")
coli_df.head()
```

```python
coli_df.columns
```

## Normalize column names

```python
coli_df.columns = (
    coli_df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
```

```python
coli_df.head()
```

```python
coli_df = coli_df[["state", "costoflivingindexbystate_2025"]]
```

```python
coli_df = coli_df.rename(columns={
    "state": "state_province",
    "costoflivingindexbystate_2025": "coli_index"
})
```

```python
coli_df.head()
```

```python
coli_df["state_province"] = coli_df["state_province"].str.strip()
coli_df["coli_index"] = pd.to_numeric(coli_df["coli_index"], errors="coerce")
```

# Generate Master dataset for the USA

```python
df_master = df.merge(
    coli_df,
    on="state_province",
    how="left"
)
```

```python
df_master[df_master["coli_index"].isna()]["state_province"].unique()
```

```python
df_master.head()
```

```python
df_master.isnull().sum()
```

Since the Cost of Living Index dataset only covers U.S. states, records from other countries (e.g., Canada) were excluded to ensure consistency in the regional efficiency analysis.

```python
df_master = df_master[df_master["country_region"] == "United States"].copy()
```

```python
df_master.isnull().sum()
```

```python
df_master["country_region"].unique()
```

## Drop Country Region since everything is the USA

```python
df_master = df_master.drop(columns=["country_region"])
```

```python
df_master.head()
```

## Final cleanning on master

```python
cols_to_drop = [
    "customer_id",
    "shipping_gap_days",
]

df_master = df_master.drop(columns=cols_to_drop)
```

```python
df_master.head()
```

# Create first CSV Table: dim_products

```python
dim_products = (
    df_master[["product_name", "division"]]
    .drop_duplicates()
    .reset_index(drop=True)
)
```

```python
# Create ID
dim_products["product_id"] = dim_products.index + 1
```

```python
# Reorder columns
dim_products = dim_products[
    ["product_id", "product_name", "division"]
]
```

# Create second CSV Table: dim_locations

```python
dim_locations = (
    df_master[["city", "state_province", "factory", "coli_index"]]
    .drop_duplicates()
    .reset_index(drop=True)
)
```

```python
# Create ID
dim_locations["location_id"] = dim_locations.index + 1
```

```python
# Rename for clarity
dim_locations = dim_locations.rename(columns={
    "factory": "factory_name"
})
```

```python
# Final order
dim_locations = dim_locations[
    ["location_id", "city", "state_province", "factory_name", "coli_index"]
]
```

# Mapping Dimension IDs to the Master Dataset

In this step, we connect the cleaned master dataset with the dimension tables 
(`dim_products` and `dim_locations`) by assigning their corresponding IDs.

## dim_products

```python
df_master = df_master.merge(
    dim_products,
    on=["product_name", "division"],
    how="left"
)
```

## dim_locations

```python
df_master = df_master.merge(
    dim_locations,
    left_on=["city", "state_province", "factory", "coli_index"],
    right_on=["city", "state_province", "factory_name", "coli_index"],
    how="left"
)
```

# Create last CSV Table: fact_sales

```python
fact_sales = df_master[[
    "product_id",
    "location_id",
    "sales",
    "cost",
    "gross_profit",
    "units"
]].copy()
```

```python
# Create order_id
fact_sales.insert(0, "order_id", range(1, len(fact_sales) + 1))
```

```python
fact_sales.isnull().sum()
```

# Export clean csv files

```python
from pathlib import Path

Path("data/clean").mkdir(parents=True, exist_ok=True)

dim_products.to_csv("data/clean/dim_products.csv", index=False)
dim_locations.to_csv("data/clean/dim_locations.csv", index=False)
fact_sales.to_csv("data/clean/fact_sales.csv", index=False)
df_master.to_csv("data/clean/wonka_master_enriched.csv", index=False)
```

# Generate SQLITE

In this step, the cleaned and structured tables are loaded into a SQLite database file (.db). 
This allows the team to perform SQL queries on a relational database instead of flat files.

```python
import sqlite3

conn = sqlite3.connect("wonka.db")

dim_products.to_sql("dim_products", conn, if_exists="replace", index=False)
dim_locations.to_sql("dim_locations", conn, if_exists="replace", index=False)
fact_sales.to_sql("fact_sales", conn, if_exists="replace", index=False)
```
