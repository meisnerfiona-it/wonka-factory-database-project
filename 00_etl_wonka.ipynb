## Drop unnecessary columns

```python
df = df.drop(columns=[
    "Unnamed: 0",
    "Latitude",
    "Longitude",
    "Postal Code",
    "Ship Mode",
    "Ship Date" 
])
```

```python
df.head()
```

```python
df=df.drop_duplicates()
```

```python
df.info()
```

```python
df.isnull().sum()
```

## Make Order date a date type

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
```

```python
df.info()
```

# Data wrangling

## Standardize titles

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("/", "_")
)
```

## Double check and erase if there are other string problems (invicible spaces...)

```python
text_cols = [
    "country_region",
    "city",
    "state_province",
    "division",
    "region",
    "product_name",
    "factory"
]

for col in text_cols:
    df[col] = df[col].str.strip()
```
